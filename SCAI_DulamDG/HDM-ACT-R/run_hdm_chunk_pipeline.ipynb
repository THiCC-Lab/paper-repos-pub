```python
test = hdm_actr.hdm_actr.hdm_modules['iatmodel']
print("before",test.chunk_count)
```

For using ConceptNet generated chunks as input use the pickle file generated.

```python
import pickle
chunks_dict={}

with open("../Data/preprocessed inputs/ConceptNet/all_cn_chunk.pkl","rb") as f:
    chunks_dict = pickle.load(f) 

print("before",test.chunk_count)
for val in chunks_dict:
    chunk_name = f"{val["topic"]}-{test.chunk_count}"
    chunk = hdm_actr.hdm_actr.Chunk(val, 
                                        chunk_name=chunk_name, 
                                        has_slots=True,
                                        is_dict=True,                                     
                                        check_defined=False)
    test.lock.acquire()
    test.add_chunk_to_dm(chunk)
    test.lock.release()
print("after",test.chunk_count)

```

```python
test.hrr_mat(chunk_count=test.chunk_count)
# This calls the "run-once" method in ACT-R task.lisp file
hdm_actr.actr.call_command("run-once")
```
